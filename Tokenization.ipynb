# Tokenization
Tokenization is the process of converting a text into smaller, meaningful units called tokens, such as words, subwords, or characters, to prepare the text for further analysis in natural language processing tasks. Tokenization is a critical step in many NLP tasks, including text processing, language modelling, and machine translation.
##Types of Tokenization
Tokenization can be classified into several types based on how the text is segmented. Here are some types of tokenization:


* Word Tokenization

*  Sentence Tokenization
*  Subword Tokenization

*  Character Tokenization










###Sentence Tokenization using sent_tokenize :-


```python
# Import Libraries
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
```

```python
# sample text
text = "I love NLP. It's fascinating! Sentence tokenization is very useful"
```

```python
#Sentence Tokenization
sentences = sent_tokenize(text)
```

```python
print(sentences)
```

### Word Tokenization using work_tokenize :-


```python
from nltk.tokenize import word_tokenize
```

```python
text = "Tokenization is the process of converting a text into smaller, meaningful units called tokens"
```

```python
words = word_tokenize(text)
```

```python
print(words)
```

#### Method 2:

```python
from nltk.tokenize import word_tokenize
```

```python
text = "Character tokenization is easy with NLTK"
```

```python
words = word_tokenize(text)
```

```python
char_token = [list(word) for word in words]
```

```python
print(char_token)
```
